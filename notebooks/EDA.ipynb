```python
# Missing values remove
df = df.dropna()

# Duplicate rows remove
df = df.drop_duplicates()

# Check shape
df.shape
```

```python
df = df[["Title", "Author", "Description", "Genres", "Rating"]]

df.head()
```

```python
df["tags"] = (
    df["Description"].astype(str) + " " +
    df["Genres"].astype(str) + " " +
    df["Author"].astype(str)
)

df[["Title", "tags"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(df["tags"])

print(tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)

print(similarity.shape)
```

```python
import pandas as pd
import joblib

# Load data
df = pd.read_csv("data/books.csv", sep="\t", header=None)

# Load similarity matrix
similarity = joblib.load("models/recommender.pkl")


def recommend(book_name):
    matches = df[df[1].str.lower() == book_name.lower()]

    if matches.empty:
        return ["Book not found"]

    index = matches.index[0]

    distances = list(enumerate(similarity[index]))
    distances = sorted(distances, key=lambda x: x[1], reverse=True)

    recommendations = []

    for i in distances[1:6]:
        recommendations.append(df.iloc[i[0], 1])

    return recommendations
```

```python
import joblib

joblib.dump(df, "../models/books.pkl")
joblib.dump(similarity, "../models/recommender.pkl")

print("Models Saved Successfully!")
```
